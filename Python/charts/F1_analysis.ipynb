## 6. Data Preparation and Table Relationships

The Formula 1 datasets are connected using common identifier columns.

For the main driver and constructor performance analysis, the following
relationships are used:

- `driverId` connects Drivers with Results
- `raceId` connects Results with Races
- `constructorId` connects Results with Constructors

### 6.1 Create Driver Name

```python
drivers["Driver Name"] = (
    drivers["forename"] + " " + drivers["surname"]
)

drivers[["driverId", "Driver Name"]].head()
```

### 6.2 Merge Drivers with Race Results

The Drivers and Results datasets are merged using `driverId` to connect
each driver with their race-level performance.

```python
driver_results = pd.merge(
    drivers,
    results,
    on="driverId",
    how="inner"
)

print("Shape after merging Drivers and Results:", driver_results.shape)

driver_results.head()
```

### 6.3 Merge with Race Information

The merged dataset is then combined with the Races dataset using
`raceId` to add information such as the race year, round, and circuit.

```python
driver_results = pd.merge(
    driver_results,
    races,
    on="raceId",
    how="inner"
)

print("Shape after merging with Races:", driver_results.shape)

driver_results.head()
```

### 6.4 Merge with Constructor Information

Finally, the Constructors dataset is merged using `constructorId`.
This allows the analysis to compare driver and constructor performance.

```python
driver_results = pd.merge(
    driver_results,
    constructors,
    on="constructorId",
    how="inner"
)

print("Final analytical dataset shape:", driver_results.shape)

driver_results.head()
```

```python
print("Final Dataset Shape:")
print(driver_results.shape)

print("\nFinal Dataset Columns:")
print(driver_results.columns.tolist())
```

## 7. Business Questions and Analysis

The prepared Formula 1 dataset is used to answer business-oriented
questions related to driver performance, constructor performance,
race participation, podium finishes, points, and season-level trends.

Each question includes an analytical calculation, visualization, and
key business insight.

### Q1. Which drivers have the highest average points per race?

```python
top_avg_points = (
    driver_results
    .groupby("Driver Name")["points"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

top_avg_points
```

```python
plt.figure(figsize=(10, 6))

top_avg_points.plot(kind="bar")

plt.title("Top 10 Drivers by Average Points")
plt.xlabel("Driver Name")
plt.ylabel("Average Points")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    charts_dir / "Q1_Top_Drivers_Average_Points.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

#### Key Insight

This analysis identifies the drivers with the highest average points
per race. A higher average indicates stronger scoring performance
across the races included in the dataset.
